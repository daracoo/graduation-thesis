```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

base_path = Path('.') 

gym_members_path = base_path / 'members' / 'gym_members_exercise_tracking.csv'
program_summary_path = base_path / '600k' / 'program_summary.csv'
programs_detailed_path = base_path / '600k' / 'programs_detailed_boostcamp_kaggle.csv'


try:
    df_members = pd.read_csv(gym_members_path)
    df_summary = pd.read_csv(program_summary_path)
    df_detailed = pd.read_csv(programs_detailed_path)
    
    print("Datasets loaded successfully!")
except FileNotFoundError as e:
    print(f"Error: Could not find file. Please check your folder structure. Details: {e}")


print("\n" + "="*30)
print("GYM MEMBERS DATASET")
print("="*30)
df_members.info()
display(df_members.head())
display(df_members.describe())

print("\n" + "="*30)
print("600K PROGRAM SUMMARY")
print("="*30)
df_summary.info()
display(df_summary.head())
display(df_summary.describe())

print("\n" + "="*30)
print("600K PROGRAMS DETAILED")
print("="*30)
df_detailed.info()
display(df_detailed.head())


display(df_detailed.describe(include='all'))
```

```python
print("GYM MEMBERS NULL'S")
print(df_members.isnull().sum())
print("\n600K PROGRAM SUMMARY NULL'S")
print(df_summary.isnull().sum())
print("\n600K PROGRAMS DETAILED NULL'S")
print(df_detailed.isnull().sum())
```

```python
print("=== GYM MEMBERS ===")
print("Workout_Type:", df_members['Workout_Type'].unique())
print("Experience_Level:", sorted(df_members['Experience_Level'].unique()))
print("Gender:", df_members['Gender'].unique())

print("\n=== PROGRAM SUMMARY ===")
print("Level unique values (sample):", df_summary['level'].unique()[:10])
print("Goal unique values (sample):", df_summary['goal'].unique()[:10])
print("Equipment unique values:", df_summary['equipment'].unique())

print("\n=== PROGRAMS DETAILED ===")
print("Equipment unique values:", df_detailed['equipment'].unique())
print("Intensity range:", df_detailed['intensity'].min(), "to", df_detailed['intensity'].max())
print("Reps range (incl. negatives):", df_detailed['reps'].min(), "to", df_detailed['reps'].max())
```
