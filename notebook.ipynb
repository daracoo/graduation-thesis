# Phase 2 - Data Preprocessing & Feature Engineering

---

> **Goal:** Clean datasets, handle missing values, and transform categorical data into numerical features for machine learning.

```python
from sklearn.preprocessing import StandardScaler, MultiLabelBinarizer, LabelEncoder
from sklearn.model_selection import train_test_split
import ast


print("\n--- Gym Members: Preprocessing ---")

df_members_clean = df_members.copy()


df_members_clean["Gender"] = df_members_clean["Gender"].map({"Male": 1, "Female": 0})
print("✓ Gender encoded: Male=1, Female=0")


df_members_clean = pd.get_dummies(df_members_clean, columns=["Workout_Type"], prefix="workout")
print(f"✓ Workout_Type one-hot encoded → columns: {[c for c in df_members_clean.columns if c.startswith('workout')]}")


def bmi_category(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25.0:
        return "Normal"
    elif bmi < 30.0:
        return "Overweight"
    else:
        return "Obese"

df_members_clean["BMI_Category"] = df_members["BMI"].apply(bmi_category)
print(f"✓ BMI_Category created: {df_members_clean['BMI_Category'].value_counts().to_dict()}")


bmi_cat_map = {"Underweight": 0, "Normal": 1, "Overweight": 2, "Obese": 3}
df_members_clean["BMI_Category_Enc"] = df_members_clean["BMI_Category"].map(bmi_cat_map)


clf_drop = ["Experience_Level", "BMI_Category"]
X_clf = df_members_clean.drop(columns=clf_drop)
y_clf = df_members_clean["Experience_Level"]  # 1, 2, 3


reg_drop = ["Calories_Burned", "BMI_Category"]
X_reg = df_members_clean.drop(columns=reg_drop)
y_reg = df_members_clean["Calories_Burned"]

print(f"\n✓ Classifier — X shape: {X_clf.shape}, y classes: {sorted(y_clf.unique())}")
print(f"✓ Regressor  — X shape: {X_reg.shape}, y range: {y_reg.min():.0f} – {y_reg.max():.0f} kcal")


scaler_clf = StandardScaler()
scaler_reg = StandardScaler()

X_clf_scaled = scaler_clf.fit_transform(X_clf)
X_reg_scaled = scaler_reg.fit_transform(X_reg)
print("✓ Features scaled with StandardScaler")


X_clf_train, X_clf_test, y_clf_train, y_clf_test = train_test_split(
    X_clf_scaled, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg_scaled, y_reg, test_size=0.2, random_state=42
)

print(f"✓ Train/test split 80/20:")
print(f"   Classifier  — train: {X_clf_train.shape[0]}, test: {X_clf_test.shape[0]}")
print(f"   Regressor   — train: {X_reg_train.shape[0]}, test: {X_reg_test.shape[0]}")


X_cluster_members = df_members_clean.drop(columns=["BMI_Category", "Experience_Level"])
scaler_cluster = StandardScaler()
X_cluster_scaled = scaler_cluster.fit_transform(X_cluster_members)
print(f"✓ Clustering matrix prepared — shape: {X_cluster_scaled.shape}")



print("\n--- Program Summary: Preprocessing ---")

df_summary_clean = df_summary.copy()


before = len(df_summary_clean)
df_summary_clean = df_summary_clean.dropna(subset=["equipment", "program_length"])
print(f"✓ Dropped {before - len(df_summary_clean)} rows with nulls in equipment/program_length")


df_summary_clean["level_list"] = df_summary_clean["level"].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) else []
)
df_summary_clean["goal_list"] = df_summary_clean["goal"].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) else []
)
print("✓ level and goal parsed from string-lists to Python lists")


mlb_level = MultiLabelBinarizer()
mlb_goal  = MultiLabelBinarizer()

level_encoded = pd.DataFrame(
    mlb_level.fit_transform(df_summary_clean["level_list"]),
    columns=[f"level_{c}" for c in mlb_level.classes_],
    index=df_summary_clean.index
)
goal_encoded = pd.DataFrame(
    mlb_goal.fit_transform(df_summary_clean["goal_list"]),
    columns=[f"goal_{c.replace(' ', '_').replace('&', 'and')}" for c in mlb_goal.classes_],
    index=df_summary_clean.index
)

print(f"✓ Level binary columns: {list(level_encoded.columns)}")
print(f"✓ Goal binary columns ({len(goal_encoded.columns)} total): {list(goal_encoded.columns)[:5]} ...")


equipment_map = {
    "Full Gym": 3,
    "Garage Gym": 2,
    "Dumbbell Only": 1,
    "At Home": 0
}
df_summary_clean["equipment_enc"] = df_summary_clean["equipment"].map(equipment_map)
print(f"✓ Equipment encoded: {equipment_map}")


num_cols_summary = ["program_length", "time_per_workout", "total_exercises"]
scaler_summary = StandardScaler()
num_scaled = pd.DataFrame(
    scaler_summary.fit_transform(df_summary_clean[num_cols_summary]),
    columns=[f"{c}_scaled" for c in num_cols_summary],
    index=df_summary_clean.index
)
print(f"✓ Numerical features scaled: {num_cols_summary}")


df_programs_matrix = pd.concat([
    level_encoded,
    goal_encoded,
    df_summary_clean[["equipment_enc"]],
    num_scaled
], axis=1)

print(f"\n✓ Final program feature matrix shape: {df_programs_matrix.shape}")
print(f"   → Each of the {df_programs_matrix.shape[0]} programs is now a vector of {df_programs_matrix.shape[1]} features")


df_summary_clean = df_summary_clean.reset_index(drop=True)
df_programs_matrix = df_programs_matrix.reset_index(drop=True)



print("\n--- Programs Detailed: Preprocessing ---")

df_detailed_clean = df_detailed.copy()


df_detailed_clean["is_timed"] = df_detailed_clean["reps"] < 0
df_detailed_clean["reps_clean"] = df_detailed_clean["reps"].abs()

timed_count = df_detailed_clean["is_timed"].sum()
print(f"✓ Timed exercises flagged: {timed_count:,} rows ({timed_count/len(df_detailed_clean)*100:.1f}%)")
print(f"  For timed exercises, reps_clean = duration in seconds")


before = len(df_detailed_clean)
df_detailed_clean = df_detailed_clean.dropna(subset=["equipment", "program_length"])
print(f"✓ Dropped {before - len(df_detailed_clean)} rows with nulls in equipment/program_length")


print("\n" + "=" * 50)
print("PHASE 2 COMPLETE — Objects ready for Phase 3 & 4:")
print("=" * 50)
print(f"  df_members_clean     → {df_members_clean.shape}  (gym members, encoded)")
print(f"  X_clf_train/test     → classifier inputs")
print(f"  X_reg_train/test     → regressor inputs")
print(f"  X_cluster_scaled     → clustering input (gym members)")
print(f"  df_summary_clean     → {df_summary_clean.shape}  (programs, parsed)")
print(f"  df_programs_matrix   → {df_programs_matrix.shape}  (program feature vectors)")
print(f"  mlb_level, mlb_goal  → fitted binarizers (needed at recommendation time)")
print(f"  scaler_clf, scaler_reg, scaler_summary → fitted scalers (needed at recommendation time)")
print(f"  df_detailed_clean    → {df_detailed_clean.shape} (exercise entries)")
```

# Phase 3 - ML Models (Gym Members Dataset)

---

> **Goal:** Train predictive models for Fitness Level Classification and Calorie Burn Regression, and perform User Clustering to identify fitness personas.

```python
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report,
    confusion_matrix, ConfusionMatrixDisplay,
    mean_squared_error, mean_absolute_error, r2_score
)
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from xgboost import XGBClassifier, XGBRegressor
import warnings
warnings.filterwarnings("ignore")

print("=" * 50)
print("PHASE 3 — ML MODELS ON GYM MEMBERS DATASET")
print("=" * 50)



print("\n--- Model 1: Fitness Level Classifier ---")


y_clf_train_xgb = y_clf_train - 1
y_clf_test_xgb  = y_clf_test  - 1

models_clf = {
    "Random Forest": RandomForestClassifier(
        n_estimators=200, max_depth=None,
        random_state=42, class_weight="balanced"
    ),
    "XGBoost": XGBClassifier(
        n_estimators=200, learning_rate=0.1,
        max_depth=5, random_state=42,
        eval_metric="mlogloss", verbosity=0
    )
}

clf_results = {}

for name, model in models_clf.items():
    if name == "XGBoost":
        model.fit(X_clf_train, y_clf_train_xgb)
        y_pred = model.predict(X_clf_test) + 1  
    else:
        model.fit(X_clf_train, y_clf_train)
        y_pred = model.predict(X_clf_test)

    acc = accuracy_score(y_clf_test, y_pred)
    f1  = f1_score(y_clf_test, y_pred, average="weighted")
    clf_results[name] = {"model": model, "preds": y_pred, "acc": acc, "f1": f1}

    print(f"\n  {name}:")
    print(f"    Accuracy : {acc:.4f}")
    print(f"    F1 Score : {f1:.4f}")


best_clf_name = max(clf_results, key=lambda k: clf_results[k]["f1"])
best_clf      = clf_results[best_clf_name]["model"]
best_clf_pred = clf_results[best_clf_name]["preds"]
print(f"\n✓ Best classifier: {best_clf_name}")


print(f"\n  Full Classification Report ({best_clf_name}):")
print(classification_report(
    y_clf_test, best_clf_pred,
    target_names=["Beginner", "Intermediate", "Advanced"]
))


fig, ax = plt.subplots(figsize=(7, 5))
cm = confusion_matrix(y_clf_test, best_clf_pred)
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Beginner", "Intermediate", "Advanced"]
)
disp.plot(ax=ax, colorbar=False, cmap="Blues")
ax.set_title(f"Confusion Matrix — {best_clf_name} Classifier", fontsize=13, pad=12)
plt.tight_layout()
savefig("6_confusion_matrix_classifier")


print("\n  Plotting feature importance...")
if best_clf_name == "Random Forest":
    importances = best_clf.feature_importances_
elif best_clf_name == "XGBoost":
    importances = best_clf.feature_importances_

feature_names = X_clf.columns.tolist()
feat_imp = pd.Series(importances, index=feature_names).sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(9, 7))
feat_imp.plot(kind="barh", ax=ax, color="#4C72B0")
ax.set_title(f"Feature Importance — {best_clf_name} Classifier", fontsize=13, pad=12)
ax.set_xlabel("Importance Score", fontsize=11)
plt.tight_layout()
savefig("7_feature_importance_classifier")




print("\n--- Model 2: Calorie Burn Predictor ---")

models_reg = {
    "Linear Regression": LinearRegression(),
    "Random Forest":     RandomForestRegressor(
        n_estimators=200, max_depth=None, random_state=42
    ),
    "XGBoost":           XGBRegressor(
        n_estimators=200, learning_rate=0.1,
        max_depth=5, random_state=42, verbosity=0
    )
}

reg_results = {}

for name, model in models_reg.items():
    model.fit(X_reg_train, y_reg_train)
    y_pred = model.predict(X_reg_test)

    rmse = np.sqrt(mean_squared_error(y_reg_test, y_pred))
    mae  = mean_absolute_error(y_reg_test, y_pred)
    r2   = r2_score(y_reg_test, y_pred)
    reg_results[name] = {"model": model, "preds": y_pred, "rmse": rmse, "mae": mae, "r2": r2}

    print(f"\n  {name}:")
    print(f"    RMSE : {rmse:.2f} kcal")
    print(f"    MAE  : {mae:.2f} kcal")
    print(f"    R²   : {r2:.4f}")


best_reg_name = max(reg_results, key=lambda k: reg_results[k]["r2"])
best_reg      = reg_results[best_reg_name]["model"]
best_reg_pred = reg_results[best_reg_name]["preds"]
print(f"\n✓ Best regressor: {best_reg_name}")


fig, axes = plt.subplots(1, 3, figsize=(13, 5))
metrics    = ["rmse", "mae", "r2"]
titles     = ["RMSE (lower is better)", "MAE (lower is better)", "R² (higher is better)"]
colors     = ["#DD8452", "#55A868", "#4C72B0"]
model_names = list(reg_results.keys())

for ax, metric, title, color in zip(axes, metrics, titles, colors):
    values = [reg_results[m][metric] for m in model_names]
    bars = ax.bar(model_names, values, color=color, alpha=0.85, edgecolor="white")
    ax.set_title(title, fontsize=11)
    ax.set_xticks(range(len(model_names)))
    ax.set_xticklabels(model_names, rotation=15, ha="right", fontsize=9)
    for bar, val in zip(bars, values):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01 * max(values),
                f"{val:.2f}", ha="center", va="bottom", fontsize=9, fontweight="bold")

fig.suptitle("Regression Model Comparison", fontsize=13, y=1.01)
plt.tight_layout()
savefig("8_regression_model_comparison")


fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(y_reg_test, best_reg_pred, alpha=0.5, color="#4C72B0", edgecolors="none", s=40)
lims = [min(y_reg_test.min(), best_reg_pred.min()) - 50,
        max(y_reg_test.max(), best_reg_pred.max()) + 50]
ax.plot(lims, lims, "r--", linewidth=1.5, label="Perfect prediction")
ax.set_xlabel("Actual Calories Burned", fontsize=12)
ax.set_ylabel("Predicted Calories Burned", fontsize=12)
ax.set_title(f"Predicted vs Actual Calories — {best_reg_name}", fontsize=13, pad=12)
ax.legend(fontsize=10)
plt.tight_layout()
savefig("9_predicted_vs_actual_calories")




print("\n--- Model 3: User Clustering (K-Means) ---")


inertias = []
sil_scores = []
K_range = range(2, 10)

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_cluster_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_cluster_scaled, km.labels_))


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

ax1.plot(list(K_range), inertias, marker="o", color="#4C72B0", linewidth=2)
ax1.set_title("Elbow Method — Inertia vs K", fontsize=12)
ax1.set_xlabel("Number of Clusters (K)")
ax1.set_ylabel("Inertia")

ax2.plot(list(K_range), sil_scores, marker="s", color="#DD8452", linewidth=2)
ax2.set_title("Silhouette Score vs K", fontsize=12)
ax2.set_xlabel("Number of Clusters (K)")
ax2.set_ylabel("Silhouette Score")

plt.suptitle("Optimal K Selection for K-Means", fontsize=13, y=1.01)
plt.tight_layout()
savefig("10_kmeans_optimal_k")


print("\n  K | Silhouette Score")
print("  " + "-" * 22)
for k, s in zip(K_range, sil_scores):
    print(f"  {k} | {s:.4f}")


FINAL_K = 3
print(f"\n✓ Overriding to K={FINAL_K} for cleaner, more interpretable personas.")

kmeans_final = KMeans(n_clusters=FINAL_K, random_state=42, n_init=10)
cluster_labels = kmeans_final.fit_predict(X_cluster_scaled)
df_members_clean["Cluster"] = cluster_labels


profile_cols = [
    "Age", "BMI", "Fat_Percentage", "Calories_Burned",
    "Session_Duration (hours)", "Workout_Frequency (days/week)",
    "Experience_Level", "Resting_BPM"
]
cluster_profile = df_members_clean.groupby("Cluster")[profile_cols].mean().round(2)
print(f"\n  Cluster Profiles (K={FINAL_K}):")
print(cluster_profile.to_string())


persona_map = {
    0: "Elite Athlete",
    1: "Lean Casual",
    2: "High-BMI Beginner"
}

df_members_clean["Persona"] = df_members_clean["Cluster"].map(persona_map)

print("\n✓ Persona distribution:")
print(df_members_clean["Persona"].value_counts())


pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_cluster_scaled)
explained = pca.explained_variance_ratio_

fig, ax = plt.subplots(figsize=(9, 7))
palette = sns.color_palette("Set2", FINAL_K)

for i in range(FINAL_K):
    mask = cluster_labels == i
    ax.scatter(
        X_pca[mask, 0], X_pca[mask, 1],
        color=palette[i], label=f"Cluster {i}",
        alpha=0.7, s=50, edgecolors="none"
    )

ax.set_title(
    f"User Clusters — PCA 2D Projection (K={FINAL_K})\n"
    f"Explained variance: PC1={explained[0]:.1%}, PC2={explained[1]:.1%}",
    fontsize=13, pad=12
)
ax.set_xlabel(f"PC1 ({explained[0]:.1%} variance)")
ax.set_ylabel(f"PC2 ({explained[1]:.1%} variance)")
ax.legend(title="Cluster", fontsize=10)
plt.tight_layout()
savefig("11b_kmeans_pca_k3")


print("\n" + "=" * 50)
print("PHASE 3 COMPLETE")
print("=" * 50)
print("Saved objects for Phase 4:")
print("  best_clf, best_clf_name  → fitness level classifier")
print("  best_reg, best_reg_name  → calorie regressor")
print("  scaler_clf, scaler_reg   → fitted scalers")
print("  X_clf.columns            → feature names for classifier input")
print("  X_reg.columns            → feature names for regressor input")
```

# Phase 4 - Recommendation Engine (600K Dataset)

> **Goal:** Cluster exercise programs and build a Content-Based Recommender using cosine similarity to match user profiles with the ideal workout plans.

```python
from sklearn.metrics.pairwise import cosine_similarity
import joblib

print("=" * 50)
print("PHASE 4 — RECOMMENDATION ENGINE")
print("=" * 50)


print("\n--- Step 1: Program Clustering ---")

inertias_prog   = []
sil_scores_prog = []
K_range_prog    = range(2, 12)

for k in K_range_prog:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(df_programs_matrix)
    inertias_prog.append(km.inertia_)
    sil_scores_prog.append(silhouette_score(df_programs_matrix, km.labels_))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

ax1.plot(list(K_range_prog), inertias_prog, marker="o", color="#4C72B0", linewidth=2)
ax1.set_title("Elbow Method — Program Clusters", fontsize=12)
ax1.set_xlabel("Number of Clusters (K)")
ax1.set_ylabel("Inertia")

ax2.plot(list(K_range_prog), sil_scores_prog, marker="s", color="#DD8452", linewidth=2)
ax2.set_title("Silhouette Score — Program Clusters", fontsize=12)
ax2.set_xlabel("Number of Clusters (K)")
ax2.set_ylabel("Silhouette Score")

plt.suptitle("Optimal K Selection — Program Clustering", fontsize=13, y=1.01)
plt.tight_layout()
savefig("12_program_kmeans_optimal_k")

print("\n  K | Silhouette Score")
print("  " + "-" * 22)
for k, s in zip(K_range_prog, sil_scores_prog):
    print(f"  {k} | {s:.4f}")

best_k_prog = K_range_prog[np.argmax(sil_scores_prog)]
print(f"\n✓ Best K by silhouette: {best_k_prog}")

kmeans_programs = KMeans(n_clusters=best_k_prog, random_state=42, n_init=10)
program_cluster_labels = kmeans_programs.fit_predict(df_programs_matrix)
df_summary_clean["Program_Cluster"] = program_cluster_labels

program_cluster_names = {
    0: "Long Heavy Programs (12w, ~74min, ~348 exercises)",
    1: "Short Light Programs (5w, ~65min, ~122 exercises)"
}

df_summary_clean["Program_Cluster_Label"] = df_summary_clean["Program_Cluster"].map(
    program_cluster_names
)

print("\n✓ Program Cluster Labels assigned:")
print(df_summary_clean["Program_Cluster_Label"].value_counts())

prog_profile_cols = ["program_length", "time_per_workout", "total_exercises"]
prog_level_cols   = [c for c in df_summary_clean.columns if c.startswith("level_")]

prog_profile = df_summary_clean.groupby("Program_Cluster")[prog_profile_cols].mean().round(2)

def dominant_value(series):
    return series.value_counts().index[0] if len(series) > 0 else "Unknown"

for col in ["level_list", "goal_list"]:
    exploded = df_summary_clean.explode(col)
    dominant = exploded.groupby("Program_Cluster")[col].agg(dominant_value)
    prog_profile[f"dominant_{col.replace('_list','')}"] = dominant

prog_profile["program_count"] = df_summary_clean.groupby("Program_Cluster").size()

print("\n  Program Cluster Profiles:")
print(prog_profile.to_string())

pca_prog = PCA(n_components=2, random_state=42)
X_prog_pca = pca_prog.fit_transform(df_programs_matrix)
explained_prog = pca_prog.explained_variance_ratio_

fig, ax = plt.subplots(figsize=(10, 7))
palette_prog = sns.color_palette("tab10", best_k_prog)

for i in range(best_k_prog):
    mask = program_cluster_labels == i
    ax.scatter(
        X_prog_pca[mask, 0], X_prog_pca[mask, 1],
        color=palette_prog[i], label=f"Cluster {i}",
        alpha=0.6, s=40, edgecolors="none"
    )

ax.set_title(
    f"Program Clusters — PCA 2D Projection (K={best_k_prog})\n"
    f"Explained variance: PC1={explained_prog[0]:.1%}, PC2={explained_prog[1]:.1%}",
    fontsize=13, pad=12
)
ax.set_xlabel(f"PC1 ({explained_prog[0]:.1%} variance)")
ax.set_ylabel(f"PC2 ({explained_prog[1]:.1%} variance)")
ax.legend(title="Program Cluster", fontsize=9, bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
savefig("13_program_clusters_pca")



print("\n--- Step 2: Content-Based Recommender ---")

cosine_sim_matrix = cosine_similarity(df_programs_matrix)
print(f"✓ Cosine similarity matrix computed: {cosine_sim_matrix.shape}")

def build_user_program_vector(goal, equipment, program_length, time_per_workout,
                               predicted_level, total_exercises_hint=None):
    level_map = {
        "Beginner":     ["Beginner", "Novice"],
        "Intermediate": ["Intermediate", "Novice"],
        "Advanced":     ["Advanced", "Intermediate"]
    }
    active_levels = level_map.get(predicted_level, ["Beginner", "Novice"])
    level_vec = {f"level_{c}": 1 if c in active_levels else 0
                 for c in mlb_level.classes_}

    goal_classes = mlb_goal.classes_
    goal_vec = {}
    for c in goal_classes:
        col_name = f"goal_{c.replace(' ', '_').replace('&', 'and')}"
        goal_vec[col_name] = 1 if c == goal else 0

    equipment_map = {"Full Gym": 3, "Garage Gym": 2, "Dumbbell Only": 1, "At Home": 0}
    equip_enc = equipment_map.get(equipment, 0)

    if total_exercises_hint is None:
        total_exercises_hint = df_summary_clean["total_exercises"].median()

    num_raw = np.array([[program_length, time_per_workout, total_exercises_hint]])
    num_scaled_vec = scaler_summary.transform(num_raw)[0]

    row = {}
    row.update(level_vec)
    row.update(goal_vec)
    row["equipment_enc"] = equip_enc
    row["program_length_scaled"]    = num_scaled_vec[0]
    row["time_per_workout_scaled"]  = num_scaled_vec[1]
    row["total_exercises_scaled"]   = num_scaled_vec[2]

    user_vec = pd.DataFrame([row])[df_programs_matrix.columns]
    return user_vec


def recommend_programs(goal, equipment, program_length, time_per_workout,
                       predicted_level, top_n=5):
    user_vec = build_user_program_vector(
        goal, equipment, program_length, time_per_workout, predicted_level
    )

    sim_scores = cosine_similarity(user_vec, df_programs_matrix)[0]
    top_indices = np.argsort(sim_scores)[::-1][:top_n]

    results = df_summary_clean.iloc[top_indices][
        ["title", "description", "level", "goal", "equipment",
         "program_length", "time_per_workout", "total_exercises"]
    ].copy()
    results["similarity_score"] = sim_scores[top_indices].round(4)
    results = results.reset_index(drop=True)
    results.index += 1
    return results



def get_week1_day1(program_title, max_exercises=10, max_timed_seconds=600):
    """
    Retrieves Week 1 Day 1 exercises for a given program title.
    Filters out corrupted timed entries over max_timed_seconds.
    600 seconds = 10 minutes, a realistic ceiling for any single set.
    """
    mask = (
        (df_detailed_clean["title"] == program_title) &
        (df_detailed_clean["week"] == 1.0) &
        (df_detailed_clean["day"] == 1.0)
    )
    exercises = df_detailed_clean[mask][
        ["exercise_name", "sets", "reps_clean", "is_timed", "intensity"]
    ].copy()

    if exercises.empty:
        return None


    valid_mask = ~((exercises["is_timed"]) & (exercises["reps_clean"] > max_timed_seconds))
    exercises = exercises[valid_mask]


    exercises = exercises[exercises["reps_clean"] > 0]

    if exercises.empty:
        return None

    exercises = exercises.head(max_exercises).reset_index(drop=True)
    exercises.index += 1

    exercises["prescription"] = exercises.apply(
        lambda r: f"{int(r['sets'])} sets × {int(r['reps_clean'])}s (timed)"
                  if r["is_timed"]
                  else f"{int(r['sets'])} sets × {int(r['reps_clean'])} reps",
        axis=1
    )
    return exercises[["exercise_name", "prescription", "intensity"]]


print("✓ recommend_programs() function ready")
print("✓ get_week1_day1() function ready")



print("\n--- Step 3: Full Recommendation Pipeline ---")

def recommend(age, weight_kg, height_m, gender, workout_type,
              workout_frequency, session_duration, fat_percentage,
              water_intake, max_bpm, avg_bpm, resting_bpm,
              goal, equipment, program_length, time_per_workout,
              top_n=5):

    print("\n" + "=" * 55)
    print("  FITNESS RECOMMENDATION SYSTEM")
    print("=" * 55)

    gender_enc       = 1 if gender == "Male" else 0
    bmi              = weight_kg / (height_m ** 2)
    bmi_cat          = 0 if bmi < 18.5 else 1 if bmi < 25 else 2 if bmi < 30 else 3

    workout_cardio   = 1 if workout_type == "Cardio"   else 0
    workout_hiit     = 1 if workout_type == "HIIT"     else 0
    workout_strength = 1 if workout_type == "Strength" else 0
    workout_yoga     = 1 if workout_type == "Yoga"     else 0

    feature_row = pd.DataFrame([{
        "Age":                           age,
        "Gender":                        gender_enc,
        "Weight (kg)":                   weight_kg,
        "Height (m)":                    height_m,
        "Max_BPM":                       max_bpm,
        "Avg_BPM":                       avg_bpm,
        "Resting_BPM":                   resting_bpm,
        "Session_Duration (hours)":      session_duration,
        "Calories_Burned":               0,
        "Fat_Percentage":                fat_percentage,
        "Water_Intake (liters)":         water_intake,
        "Workout_Frequency (days/week)": workout_frequency,
        "BMI":                           bmi,
        "workout_Cardio":                workout_cardio,
        "workout_HIIT":                  workout_hiit,
        "workout_Strength":              workout_strength,
        "workout_Yoga":                  workout_yoga,
        "BMI_Category_Enc":              bmi_cat
    }])

    if "Experience_Level" in X_clf.columns:
        feature_row["Experience_Level"] = 2

    feature_row_clf = feature_row[X_clf.columns]

    feature_scaled_clf = scaler_clf.transform(feature_row_clf)
    level_num = int(best_clf.predict(feature_scaled_clf)[0]) + 1
    level_map_str = {1: "Beginner", 2: "Intermediate", 3: "Advanced"}
    predicted_level = level_map_str.get(level_num, "Intermediate")

    feature_row_reg = feature_row.copy()
    feature_row_reg["Experience_Level"] = level_num
    if "Calories_Burned" in feature_row_reg.columns:
        feature_row_reg = feature_row_reg.drop(columns=["Calories_Burned"])
    feature_row_reg = feature_row_reg[X_reg.columns]
    feature_scaled_reg = scaler_reg.transform(feature_row_reg)
    predicted_calories = best_reg.predict(feature_scaled_reg)[0]

    recommendations = recommend_programs(
        goal=goal,
        equipment=equipment,
        program_length=program_length,
        time_per_workout=time_per_workout,
        predicted_level=predicted_level,
        top_n=top_n
    )

    top_program_title = recommendations.iloc[0]["title"]
    week1_day1 = get_week1_day1(top_program_title)

    print(f"\n  📋 USER PROFILE SUMMARY")
    print(f"  {'─'*45}")
    print(f"  Age: {age}  |  Gender: {gender}  |  BMI: {bmi:.1f}")
    print(f"  Predicted Fitness Level : {predicted_level}")
    print(f"  Estimated Calorie Burn  : {predicted_calories:.0f} kcal/session")
    print(f"  Goal: {goal}  |  Equipment: {equipment}")
    print(f"  Program Length: {program_length} weeks  |  {time_per_workout} min/session")

    print(f"\n  🏆 TOP {top_n} RECOMMENDED PROGRAMS")
    print(f"  {'─'*45}")
    for idx, row in recommendations.iterrows():
        print(f"  {idx}. {row['title']}")
        print(f"     Goal: {row['goal']}  |  Level: {row['level']}")
        print(f"     {row['program_length']:.0f} weeks  |  "
              f"{row['time_per_workout']:.0f} min/session  |  "
              f"{row['total_exercises']} exercises total")
        print(f"     Similarity: {row['similarity_score']:.4f}")
        print()

    if week1_day1 is not None:
        print(f"  📅 WEEK 1 · DAY 1 — '{top_program_title}'")
        print(f"  {'─'*45}")
        for i, ex in week1_day1.iterrows():
            print(f"  {i:>2}. {ex['exercise_name']:<35} {ex['prescription']}")
    else:
        print(f"  ℹ️  Week 1 Day 1 details not available for this program.")

    print("\n" + "=" * 55)

    return {
        "predicted_level":    predicted_level,
        "predicted_calories": round(predicted_calories, 1),
        "bmi":                round(bmi, 2),
        "recommendations":    recommendations,
        "week1_day1":         week1_day1
    }


print("✓ recommend() pipeline function ready")
print("\n" + "=" * 50)
print("PHASE 4 COMPLETE — Running test examples...")
print("=" * 50)


# ─────────────────────────────────────────────────────
# TEST EXAMPLES
# ─────────────────────────────────────────────────────

# Example 1 — Young beginner, weight loss, home workout
result1 = recommend(
    age=24, weight_kg=92.0, height_m=1.75, gender="Male",
    workout_type="Cardio", workout_frequency=3, session_duration=0.75,
    fat_percentage=28.0, water_intake=2.0, max_bpm=175, avg_bpm=140,
    resting_bpm=68, goal="Weight Loss", equipment="At Home",
    program_length=8, time_per_workout=45
)

# Example 2 — Experienced female, muscle building, full gym
result2 = recommend(
    age=32, weight_kg=62.0, height_m=1.65, gender="Female",
    workout_type="Strength", workout_frequency=5, session_duration=1.5,
    fat_percentage=18.0, water_intake=3.0, max_bpm=185, avg_bpm=155,
    resting_bpm=55, goal="Bodybuilding", equipment="Full Gym",
    program_length=12, time_per_workout=90
)

# Example 3 — Middle-aged male, general fitness, dumbbells only
result3 = recommend(
    age=45, weight_kg=78.0, height_m=1.80, gender="Male",
    workout_type="HIIT", workout_frequency=4, session_duration=1.0,
    fat_percentage=22.0, water_intake=2.5, max_bpm=178, avg_bpm=148,
    resting_bpm=62, goal="Muscle & Sculpting", equipment="Dumbbell Only",
    program_length=10, time_per_workout=60
)
```

# Phase 5 - Evaluation & Results

---

> **Goal:** Evaluate the recommendation engine's performance on real-world test cases, synthesize the consolidated thesis results, extract key behavioral findings, and serialize all models for production readiness.

```python
import joblib
import os
from pathlib import Path 

print("=" * 50)
print("PHASE 5 — EVALUATION & RESULTS")
print("=" * 50)

# ─────────────────────────────────────────────────────
# PART A — RECOMMENDER EVALUATION
# ─────────────────────────────────────────────────────
print("\n--- Part A: Recommender Evaluation ---")

def evaluate_recommendations(result, user_goal, user_equipment,
                              user_level, user_program_length, user_time):
    """
    Manually scores each recommendation on 5 criteria:
    - Goal match (exact or partial)
    - Equipment match (exact)
    - Level match (contains predicted level)
    - Length proximity (within 3 weeks)
    - Time proximity (within 20 minutes)
    Returns relevance scores and diversity metrics.
    """
    recs = result["recommendations"]
    predicted_level = result["predicted_level"]
    scores = []

    for _, row in recs.iterrows():
        score = 0

        # Goal match — check if user goal appears in program goal list
        goal_match = user_goal.lower() in str(row["goal"]).lower()
        score += 2 if goal_match else 0

        # Equipment match
        equip_match = user_equipment.lower() in str(row["equipment"]).lower()
        score += 2 if equip_match else 0

        # Level match — check if predicted level appears in program level list
        level_match = predicted_level.lower() in str(row["level"]).lower()
        score += 2 if level_match else 0

        # Program length proximity (within 3 weeks)
        length_diff = abs(row["program_length"] - user_program_length)
        score += 2 if length_diff <= 3 else 1 if length_diff <= 6 else 0

        # Time per workout proximity (within 20 minutes)
        time_diff = abs(row["time_per_workout"] - user_time)
        score += 2 if time_diff <= 20 else 1 if time_diff <= 40 else 0

        scores.append(score)

    max_score = 10
    relevance_pct = [round(s / max_score * 100, 1) for s in scores]

    # Diversity — FIXED: Use Python built-in round() on pandas std() float outputs
    length_diversity = round(recs["program_length"].std(), 2)
    time_diversity   = round(recs["time_per_workout"].std(), 2)

    return scores, relevance_pct, length_diversity, time_diversity


# Evaluate all 3 examples
eval_cases = [
    {
        "label":          "Example 1 — Young Beginner (Weight Loss, At Home)",
        "result":         result1,
        "goal":           "Weight Loss",
        "equipment":      "At Home",
        "level":          "Beginner",
        "program_length": 8,
        "time":           45
    },
    {
        "label":          "Example 2 — Experienced Female (Bodybuilding, Full Gym)",
        "result":         result2,
        "goal":           "Bodybuilding",
        "equipment":      "Full Gym",
        "level":          "Intermediate",
        "program_length": 12,
        "time":           90
    },
    {
        "label":          "Example 3 — Middle-aged Male (Muscle & Sculpting, Dumbbells)",
        "result":         result3,
        "goal":           "Muscle & Sculpting",
        "equipment":      "Dumbbell Only",
        "level":          "Intermediate",
        "program_length": 10,
        "time":           60
    }
]

all_avg_relevance = []

for case in eval_cases:
    scores, relevance_pct, len_div, time_div = evaluate_recommendations(
        result         = case["result"],
        user_goal      = case["goal"],
        user_equipment = case["equipment"],
        user_level     = case["level"],
        user_program_length = case["program_length"],
        user_time      = case["time"]
    )
    avg_relevance = round(sum(relevance_pct) / len(relevance_pct), 1)
    all_avg_relevance.append(avg_relevance)

    print(f"\n  {case['label']}")
    print(f"  {'─' * 50}")
    for i, (s, pct) in enumerate(zip(scores, relevance_pct), 1):
        bar = "█" * int(pct // 10) + "░" * (10 - int(pct // 10))
        print(f"  Rank {i}: [{bar}] {pct}%  (raw score: {s}/10)")
    print(f"  Average Relevance : {avg_relevance}%")
    print(f"  Program Length Diversity (std): {len_div} weeks")
    print(f"  Time per Session Diversity (std): {time_div} min")

overall_relevance = round(sum(all_avg_relevance) / len(all_avg_relevance), 1)
print(f"\n✓ Overall Average Relevance across all examples: {overall_relevance}%")


# ─────────────────────────────────────────────────────
# PART B — CONSOLIDATED RESULTS SUMMARY
# ─────────────────────────────────────────────────────
print("\n--- Part B: Consolidated Results Summary ---")

print("""
╔══════════════════════════════════════════════════════════════╗
║           THESIS RESULTS — COMPLETE SUMMARY                  ║
╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  DATASET OVERVIEW                                            ║
║  • Gym Members Dataset    :   973 users, 15 features         ║
║  • Program Summary        : 2,597 programs, 10 columns       ║
║  • Programs Detailed      : 605,017 exercise entries         ║
║                                                              ║
╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  MODEL 1 — FITNESS LEVEL CLASSIFIER (XGBoost)                ║
║  • Accuracy               : 90.26%                           ║
║  • Weighted F1 Score      : 0.9025                           ║
║  • Advanced class         : 100% precision & recall          ║
║  • Top features           : Workout Frequency, Session       ║
║                             Duration, Fat Percentage         ║
║                                                              ║
╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  MODEL 2 — CALORIE BURN PREDICTOR (XGBoost)                  ║
║  • RMSE                   : 30.51 kcal                       ║
║  • MAE                    : 19.59 kcal                       ║
║  • R²                     : 0.9888                           ║
║  • Baseline (Linear Reg)  : R² = 0.9806                      ║
║                                                              ║
╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  MODEL 3 — USER CLUSTERING (K-Means, K=3)                    ║
║  • Silhouette Score       : 0.1546                           ║
║  • Cluster 0 — Elite Athlete      : 209 users (21.5%)        ║
║  • Cluster 1 — Lean Casual        : 520 users (53.4%)        ║
║  • Cluster 2 — High-BMI Beginner  : 244 users (25.1%)        ║
║                                                              ║
╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  PROGRAM CLUSTERING (K-Means, K=2)                           ║
║  • Silhouette Score       : 0.1721                           ║
║  • Cluster 0 — Long Heavy : 1,275 programs (49.1%)           ║
║  • Cluster 1 — Short Light: 1,322 programs (50.9%)           ║
║                                                              ║
╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  RECOMMENDATION ENGINE (Cosine Similarity)                   ║
║  • Programs in catalog    : 2,597                            ║
║  • Feature vector size    : 15 dimensions                    ║
║  • Best similarity score  : 0.9981 (Example 2)               ║
║  • Avg similarity (Ex 1)  : ~0.82                            ║
║  • Avg similarity (Ex 2)  : ~0.97                            ║
║  • Avg similarity (Ex 3)  : ~0.87                            ║""")
print(f"║  • Overall avg relevance  : {overall_relevance}%"
      + " " * (32 - len(str(overall_relevance))) + "║")
print("""║                                                              ║
╚══════════════════════════════════════════════════════════════╝""")


# ─────────────────────────────────────────────────────
# PART C — KEY FINDINGS SUMMARY
# ─────────────────────────────────────────────────────
print("\n--- Part C: Key Findings ---")
print("""
  1. BEHAVIORAL > PHYSICAL for fitness level classification
     Workout frequency and session duration account for ~70%
     of feature importance. Age, BMI, and heart rate metrics
     contribute very little. Experience is a behavioral trait.

  2. CALORIE BURN IS HIGHLY PREDICTABLE
     Even Linear Regression achieves R²=0.98, confirming
     a near-linear relationship between session duration,
     body metrics, and calories burned.

  3. THREE DISTINCT USER PERSONAS EXIST
     K-Means revealed Elite Athletes (21.5%), Lean Casuals
     (53.4%), and High-BMI Beginners (25.1%). Age is nearly
     identical across all clusters (~38 years), confirming
     fitness level is driven by behavior, not age.

  4. PROGRAMS SPLIT INTO TWO STRUCTURAL TYPES
     The 600K catalog naturally divides into long heavy
     programs (12w, 74min) and short light programs (5w, 65min)
     with a clear silhouette peak at K=2.

  5. GOAL VOCABULARY GAP
     "Weight Loss" does not exist in the 600K dataset's
     goal vocabulary. This real-world limitation shows
     the importance of dataset alignment in RS design.
""")


# ─────────────────────────────────────────────────────
# PART D — SAVE ALL MODELS
# ─────────────────────────────────────────────────────
print("\n--- Part D: Saving Models ---")

models_dir = Path("models")
models_dir.mkdir(exist_ok=True)

joblib.dump(best_clf,          models_dir / "classifier_xgboost.pkl")
joblib.dump(best_reg,          models_dir / "regressor_xgboost.pkl")
joblib.dump(scaler_clf,        models_dir / "scaler_classifier.pkl")
joblib.dump(scaler_reg,        models_dir / "scaler_regressor.pkl")
joblib.dump(scaler_summary,    models_dir / "scaler_summary.pkl")
joblib.dump(mlb_level,         models_dir / "mlb_level.pkl")
joblib.dump(mlb_goal,          models_dir / "mlb_goal.pkl")
joblib.dump(kmeans_final,      models_dir / "kmeans_users.pkl")
joblib.dump(kmeans_programs,   models_dir / "kmeans_programs.pkl")
joblib.dump(cosine_sim_matrix, models_dir / "cosine_similarity_matrix.pkl")

saved_files = list(models_dir.iterdir())
print(f"\n✓ {len(saved_files)} objects saved to /models:")
for f in sorted(saved_files):
    size_kb = os.path.getsize(f) / 1024
    print(f"   {f.name:<40} {size_kb:>8.1f} KB")

print("\n" + "=" * 50)
print("PHASE 5 COMPLETE")
print("=" * 50)
```
